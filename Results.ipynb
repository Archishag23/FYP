```python
import pandas as pd
import matplotlib.pyplot as plt
sectors = pd.read_excel('SPX_sectors_data.xlsx', sheet_name='Sectors', header=0, index_col=0)
results = pd.read_pickle("outputs/test_results_test_11Apr_fix2_diff2.pkl")

# --- Build per-sector mean signal time series ---
sector_signal_rows = []
stock_sector_map = sectors.iloc[:, 0]  # ticker -> sector name

for t, (stocks, signals) in results.items():
    s_signals = pd.Series(signals, index=stocks)
    aligned_sectors = stock_sector_map.reindex(s_signals.index)
    row = s_signals.groupby(aligned_sectors).mean()
    row.name = t
    sector_signal_rows.append(row)

df_sector_signals = pd.DataFrame(sector_signal_rows).sort_index()
df_sector_signals.index = pd.to_datetime(df_sector_signals.index)

# --- Rolling mean (22 trading days ~ 1 month) ---
ROLL = 22
df_rolled = df_sector_signals.rolling(ROLL, min_periods=1).mean()

# --- Known crisis windows to shade ---
crises = [
    ("COVID",    "2020-02-01", "2020-06-30"),
]

# --- Single combined plot ---
fig, ax = plt.subplots(figsize=(16, 6))

for col in df_rolled.columns:
    ax.plot(df_rolled.index, df_rolled[col], label=col, linewidth=1.3)

ymin, ymax = ax.get_ylim()
for label, start, end in crises:
    s, e = pd.Timestamp(start), pd.Timestamp(end)
    if s > df_rolled.index[-1] or e < df_rolled.index[0]:
        continue
    ax.axvspan(s, e, alpha=0.12, color="red", zorder=0)
    ax.text(s, ymax * 0.98, label, fontsize=8, color="darkred", va="top")

ax.set_title(f"Rolling {ROLL}-day Mean Anomaly Signal by Sector (crisis periods shaded)")
ax.set_ylabel("Mean Anomaly Signal")
ax.set_xlabel("Date")
ax.legend(loc="upper left", bbox_to_anchor=(1.01, 1), fontsize=9, frameon=True)
plt.tight_layout()
plt.show()

# --- Small multiples: one subplot per sector ---
sectors_list = df_rolled.columns.tolist()
n = len(sectors_list)
ncols = 3
nrows = (n + ncols - 1) // ncols

fig, axes = plt.subplots(nrows, ncols, figsize=(16, nrows * 3), sharex=True)
axes = axes.flatten()

for i, sector in enumerate(sectors_list):
    ax = axes[i]
    ax.plot(df_rolled.index, df_rolled[sector], linewidth=1.2)
    ymin, ymax = ax.get_ylim()
    for label, start, end in crises:
        s, e = pd.Timestamp(start), pd.Timestamp(end)
        if s > df_rolled.index[-1] or e < df_rolled.index[0]:
            continue
        ax.axvspan(s, e, alpha=0.15, color="red", zorder=0)
    ax.set_title(sector, fontsize=9)
    ax.tick_params(axis='x', labelrotation=30, labelsize=7)

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

fig.suptitle(f"Rolling {ROLL}-day Mean Anomaly Signal — per Sector", y=1.01)
plt.tight_layout()
plt.show()
```
